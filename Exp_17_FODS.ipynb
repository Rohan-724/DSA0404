```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the newly uploaded dataset with latin-1 encoding to prevent decoding errors
df = pd.read_csv('17customer_reviews.csv', encoding='latin-1')

# Count words from Review Text column
review = df['Review Text'].fillna('').str.lower().str.split().explode().value_counts()

# Remove common words
remove = ["the", "and", "is", "a", "to", "of", "in", "for"]
rem_w = review[~review.index.isin(remove)]

print("Word Frequencies:")
print(rem_w.head(10))

print("\nTop 5 Most Frequent Words:")
print(rem_w.nlargest(5))

# Plot top 5 words
top5 = rem_w.nlargest(5)

plt.bar(top5.index, top5.values)
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title("Most Frequent Words in Reviews")
plt.show()
```
