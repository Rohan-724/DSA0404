```python
!pip install pdfplumber

from google.colab import files

uploaded = files.upload()

pdf_file = next(iter(uploaded))

print("PDF uploaded successfully:", pdf_file)
```

```python
import pandas as pd
import pdfplumber

all_data = []

with pdfplumber.open(pdf_file) as pdf:
    for page in pdf.pages:
        text = page.extract_text()
        if text:
            for line in text.split("\n"):
                parts = line.split()
                if len(parts) == 2:
                    try:
                        hours = float(parts[0])
                        scores = float(parts[1])
                        all_data.append([hours, scores])
                    except ValueError:
                        pass

df = pd.DataFrame(all_data, columns=["Hours", "Scores"])

print("Extracted Data:")
print(df)

print("\nScore Frequency:")
print(df["Scores"].value_counts().sort_index())
```
