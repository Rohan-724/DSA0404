```python
import numpy as np

house_data = np.array([
    [3, 1500, 200000],
    [5, 2500, 350000],
    [4, 1800, 250000],
    [6, 3000, 500000]
])

avg_price = np.mean(house_data[house_data[:,0] > 4][:,2])
print("Average sale price of houses with more than four bedrooms:", avg_price)
```
