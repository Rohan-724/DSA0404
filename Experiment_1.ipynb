```python
import numpy as np

student_scores = np.array([
    [85, 78, 92, 74],
    [88, 90, 76, 85],
    [70, 65, 80, 72],
    [95, 88, 89, 90]
])

subjects = ["Math", "Science", "English", "History"]

subject_averages = np.mean(student_scores, axis=0)
highest_avg_index = np.argmax(subject_averages)
highest_subject = subjects[highest_avg_index]

print("Average scores per subject:", subject_averages)
print("Subject with highest average:", highest_subject)
```
