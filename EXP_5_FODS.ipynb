```python
import pandas as pd
from google.colab import files

uploaded = files.upload()
filename = list(uploaded.keys())[0]

if filename.endswith('.csv'):
    df = pd.read_csv(filename)
elif filename.endswith('.pdf'):
    !pip install pdfplumber -q
    import pdfplumber
    with pdfplumber.open(filename) as pdf:
        data = [p.extract_text() for p in pdf.pages]
```

```python
import numpy as np
import pandas as pd
import re


rows = []
for page_text in data:
    for line in page_text.strip().split('\n'):
        parts = line.split()
        if parts and parts[0].isdigit():
            rows.append(parts)

df = pd.DataFrame(rows)

df['city_mpg'] = pd.to_numeric(df[0], errors='coerce')
df = df.dropna(subset=['city_mpg'])

fuel_efficiency = np.array(df['city_mpg'])

print("Average Fuel Efficiency:", np.mean(fuel_efficiency))

car1 = df.iloc[0]['city_mpg']
car2 = df.iloc[1]['city_mpg']

improvement = ((car2 - car1) / car1) * 100

print("Car 1 Efficiency:", car1)
print("Car 2 Efficiency:", car2)
print("Percentage Improvement:", improvement, "%")
```
