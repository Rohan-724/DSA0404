```python
import pdfplumber
import pandas as pd

pdf_path = "/content/15.weather.pdf"

left_rows = []
right_rows = []

with pdfplumber.open(pdf_path) as pdf:
    for i, page in enumerate(pdf.pages):
        text = page.extract_text()
        if not text:
            continue

        # Split lines and ignore the headers
        lines = [line.split() for line in text.strip().split("\n")]

        if (i + 1) % 2 != 0:
            # Odd pages: Region Country State City Month Day
            for line in lines:
                if line and line[0] != "Region":
                    # Ensure the line has enough elements and ends with numeric month/day values
                    if len(line) >= 5 and line[-1].isdigit() and line[-2].isdigit():
                        month = line[-2]
                        day = line[-1]
                        city = " ".join(line[2:-2])
                        country = line[1]
                        region = line[0]
                        left_rows.append([region, country, city, int(month), int(day)])
        else:
            # Even pages: Year AvgTemperature
            for line in lines:
                if line and line[0] != "Year":
                    # Verify we have at least 2 elements and that Year is a digit
                    if len(line) >= 2 and line[0].isdigit():
                        year = line[0]
                        temp = line[1]
                        try:
                            right_rows.append([int(year), float(temp)])
                        except ValueError:
                            continue

# Combine the aligned columns safely using the shorter list size to prevent index mismatches
combined_data = []
for left, right in zip(left_rows, right_rows):
    combined_data.append(left + right)

# Create DataFrame
columns = ["Region", "Country", "City", "Month", "Day", "Year", "AvgTemperature"]
df = pd.DataFrame(combined_data, columns=columns)

# Display the structured DataFrame
display(df.head())
```

```python
# Calculate and print the frequency distribution
temperature_frequency = df["AvgTemperature"].value_counts()

print("Temperature Frequency Distribution:")
print(temperature_frequency)
```
