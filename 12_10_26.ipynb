```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample dataset
data = {
    "Student": ["Arun","Bala","Chitra","Divya","Ezhil","Farhan","Geetha","Hari"],
    "Department": ["CSE","AI","CSE","AI","IT","IT","CSE","AI"],
    "Gender": ["Male","Male","Female","Female","Male","Male","Female","Male"],
    "Study_Hours": [2,4,3,5,6,2,4,5],
    "Marks": [50,70,65,85,90,55,75,80],
    "Attendance": [70,85,80,95,90,65,88,92]
}
df = pd.DataFrame(data)

sns.set_theme(style="whitegrid")

plt.figure(figsize=(6,4))
sns.violinplot(data=df, x="Department", y="Marks")
plt.title("Violin Plot: Marks Distribution by Department")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

plt.figure(figsize=(6,4))
sns.histplot(data=df, x="Marks", bins=5, kde=True)
plt.title("Histogram: Marks Distribution")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

plt.figure(figsize=(6,4))
sns.kdeplot(data=df, x="Marks", fill=True)
plt.title("KDE Plot: Marks Density")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

sns.pairplot(df[["Study_Hours","Marks","Attendance"]])
plt.suptitle("Pair Plot: Numeric Variable Relationships", y=1.02)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

correlation = df[["Study_Hours","Marks","Attendance"]].corr()

plt.figure(figsize=(6,4))
sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Heatmap: Correlation Matrix")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

plt.figure(figsize=(6,4))
sns.regplot(data=df, x="Study_Hours", y="Marks")
plt.title("Regression Plot: Study Hours vs Marks")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

plt.figure(figsize=(6,4))
sns.stripplot(data=df, x="Department", y="Marks")
plt.title("Strip Plot: Marks by Department")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

plt.figure(figsize=(6,4))
sns.swarmplot(data=df, x="Department", y="Marks")
plt.title("Swarm Plot: Marks by Department")
plt.show()
```
